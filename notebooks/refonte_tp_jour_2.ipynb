```python
from pathlib import Path

import pandas as pd

SOURCE = Path("../data/france/food_france.csv.gz")
OUTPUT = Path("../data/sample_500.csv")

sample = pd.read_csv(
    SOURCE,
    compression="gzip",
    sep="\t",
    nrows=500,
    dtype=str,
    on_bad_lines="skip",
)

sample.to_csv(OUTPUT, index=False)
print(f"{len(sample)} lignes, {sample.shape[1]} colonnes -> {OUTPUT}")
```

```python
df = pd.read_csv(OUTPUT, dtype=str)

df.info()                # colonnes, types, valeurs non nulles
df.head()                # aperçu des premières lignes
df.isna().mean().sort_values(ascending=False).head(20)   # 20 colonnes les plus vides

df.columns.to_list()
```
